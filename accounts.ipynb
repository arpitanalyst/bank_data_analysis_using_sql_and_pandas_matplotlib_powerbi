```python
import pandas as pd

# Load dataset
transactions = pd.read_csv("transactions.csv")

# Convert date column
transactions["transaction_datetime"] = pd.to_datetime(
    transactions["transaction_datetime"]
)

# Calculate KPIs
total_transactions = transactions["transaction_id"].nunique()

total_amount = transactions["amount_inr"].sum()

average_transaction = transactions["amount_inr"].mean()

successful_transactions = (
    transactions["status"].str.strip().str.lower() == "success"
).sum()

failed_transactions = (
    transactions["status"].str.strip().str.lower() == "failed"
).sum()

failed_rate = failed_transactions / len(transactions) * 100

total_accounts = transactions["account_id"].nunique()

total_categories = transactions["category"].nunique()

most_used_type = transactions["transaction_type"].mode()[0]

top_category = (
    transactions.groupby("category")["amount_inr"]
    .sum()
    .idxmax()
)

# Display KPI results
print("Total Transactions:", total_transactions)
print("Total Transaction Amount:", round(total_amount, 2))
print("Average Transaction Value:", round(average_transaction, 2))
print("Successful Transactions:", successful_transactions)
print("Failed Transactions:", failed_transactions)
print("Failed Transaction Rate (%):", round(failed_rate, 2))
print("Total Unique Accounts:", total_accounts)
print("Total Categories:", total_categories)
print("Most Used Transaction Type:", most_used_type)
print("Top Category by Amount:", top_category)
```

```python
transactions["month"] = (
    transactions["transaction_datetime"].dt.to_period("M")
)

monthly_kpi = transactions.groupby("month").agg(
    total_transactions=("transaction_id", "nunique"),
    total_amount=("amount_inr", "sum"),
    average_amount=("amount_inr", "mean")
)

print(monthly_kpi)
```
